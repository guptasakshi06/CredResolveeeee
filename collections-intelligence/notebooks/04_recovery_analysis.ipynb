# 04. Recovery analysis
This notebook quantifies recovery performance using the cleaned event and payment tables.

```python
import pandas as pd
payments = pd.read_csv('../collections-intelligence/data/raw/payments.csv')
payments['event_at'] = pd.to_datetime(payments['event_at'], errors='coerce')
s = payments[payments['payment_status'].str.upper().eq('SUCCESS')].copy()
monthly = s.assign(month=s['event_at'].dt.to_period('M').astype(str)).groupby('month')['amount'].sum().sort_index()
print(monthly.round(2).to_dict())
```
