# 01. Data profiling
This notebook establishes the baseline dataset profile before any cleaning or metric logic is applied.
The purpose is to quantify schema health, duplicate risk, null rate, time range, and entity drift.

```python
import pandas as pd
import glob
import os
base = '../collections-intelligence/data/raw'
for path in sorted(glob.glob(os.path.join(base, '*.csv'))):
    df = pd.read_csv(path)
    print('FILE:', os.path.basename(path))
    print('rows=', len(df), 'cols=', len(df.columns))
    print('null_pct=', df.isna().mean().mul(100).round(2).to_dict())
    print('duplicates=', int(df.duplicated().sum()))
    print('---')
```
