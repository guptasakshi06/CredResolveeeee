# 05. Statistical investigation
This notebook reviews mix effects, attribution bias, and simple inference to separate operational improvement from reporting distortion.

```python
import pandas as pd
from scipy import stats
payments = pd.read_csv('../collections-intelligence/data/raw/payments.csv')
payments['amount'] = pd.to_numeric(payments['amount'], errors='coerce')
print('payment amount mean=', payments['amount'].mean())
print('payment amount std=', payments['amount'].std())
print('success rate=', (payments['payment_status'].astype(str).str.upper().eq('SUCCESS')).mean())
```
