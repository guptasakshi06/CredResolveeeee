# 02. Data forensics
This notebook investigates duplicate signals, attribution risk, and ID mismatches.

```python
import pandas as pd
payments = pd.read_csv('../collections-intelligence/data/raw/payments.csv')
payments['event_at'] = pd.to_datetime(payments['event_at'], errors='coerce')
print('duplicate payment refs=', int(payments['payment_reference'].duplicated(keep=False).sum()))
print('raw success amount=', round(payments.loc[payments['payment_status'].astype(str).str.upper().eq('SUCCESS'), 'amount'].sum(), 2))
print('deduped success amount=', round(payments.loc[payments['payment_status'].astype(str).str.upper().eq('SUCCESS')].drop_duplicates(subset='payment_reference', keep='first')['amount'].sum(), 2))
```
